```python
import nums


class Solution_1:
    def __init__(self,nums:list[int]):
        self.nums=nums
    def index_sum(self,left:int,right:int):
        total=0
        for i in range(left,right+1):
            total +=self.nums[i]
        return total
```

```python
#前缀和复杂度o(1)
class Solution_2:
    def __init__(self,nums:list[int]):
        #⚠前缀和长度数组加一
        self.preSum=[0]*(len(nums)+1)
        for i in range(0,len(nums)-1):
            self.preSum[i+1]=self.preSum[i]+ nums[i]
    def index_sum(self,left:int,right:int):
        total = self.preSum[right+1]-self.preSum[left]
        return total
```

```python
s =Solution_2([1,2,3,4,5])
print(s.index_sum(0,3))
```

```python
#给你一个整数数组 nums 和一个整数 k，请你统计并返回 该数组中和为 k 的子数组的个数。子数组，只能连续滑动窗口，但是没说是负数还是正数，滑动窗口要么全为正数要么全为负数要求和想到前缀和
#子数组是数组中元素的连续非空序列。
class Solution:
    # 前缀和+哈希表
    def subarraySum(self, nums: list[int], k: int) -> int:
        # 构造前缀和
        pre_sum=[0]*(len(nums)+1)
        for i in range(len(nums)):
            pre_sum[i+1]=pre_sum[i]+nums[i]

        cache={} # 某些值出现了几次
        count=0
        for i,item in enumerate(pre_sum):
            other=item-k
            if other in cache:
                count+=cache[other]
            cache[item]=cache.get(item,0)+1
        return count

s = Solution()
# 调用时直接传数组和数字，不要加类型标注
print(s.subarraySum([1,-2,-1,2],0))
```
