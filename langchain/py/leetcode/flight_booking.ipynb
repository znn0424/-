```python
#这里有 n 个航班，它们分别从 1 到 n 进行编号。
#有一份航班预订表 bookings ，表中第 i 条预订记录 bookings [i] = [firstᵢ, lastᵢ, seatsᵢ] 意味着在从 firstᵢ 到 lastᵢ（包含 firstᵢ 和 lastᵢ）的每个航班上预订了 seatsᵢ 个座位。
#请你返回一个长度为 n 的数组 answer，里面的元素是每个航班预定的座位总数。
#差分数组：在区间内累加累减，记录变换即可
class Solution:
    def flight_booking(self,n:int,bookings:list[list[int]]):
        diff =[0]*(n+1)
        last_seat=0
        answer =[0]*n
        for first,last,seat in bookings:
          diff[first-1]+= seat
          diff[last]-= seat
        for i in range(n):
            answer[i]=last_seat+diff[i]
            last_seat =answer[i]
        return answer
s = Solution()
print(s.flight_booking(4,[[1,3,4],[2,4,3]]))

```
